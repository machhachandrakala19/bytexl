#### conditional statements

```python
narr=np.array([10,15,25,30,45,50])
res=np.where(narr<20,"small",np.where(narr<40,"Medium","large"))
print(res)
```

#### Numpy Select function

```python
narr =np.array([10,45,25,30,45,50])
cons=[narr<20,(narr>=20)&(narr<40),narr>=40]
sel=["small","medium","large"]
res=np.select(cons,sel,default="Unknown")
print(res)
```

```python
import pandas as pd
data={
    "players":["virat","rohit","kohli","dhoni","sehwag","dhoni","virat","rohit"],
    "matches":[100,200,300,400,100,200,300,400],
    "runs":[154,200,123,257,309,198,171,120],
    "Balls":[50,100,120,120,150,80,120,100]
}
ipl=pd.DataFrame(data)
ipl
```

```python
ipl['strike rate']=(ipl['runs']/ipl['Balls'])*100
print(ipl)
fil_ipl=ipl[(ipl['players']=='virat')&(ipl['runs']>150)]
print(fil_ipl)

```

```python
player_stats = ipl.groupby('players')['runs'].agg(Total_Runs='sum', Highest_Score='max').reset_index()

print(player_stats)
```

```python
import pandas as pd
data={
    "players":["virat","rohit","kohli","dhoni","sehwag","dhoni","virat","rohit"],
    "matches":[100,200,300,400,100,200,300,400],
    "runs":[154,200,123,257,309,198,171,54],
    "Balls":[50,100,120,120,150,80,120,100]
}
df=pd.DataFrame(data)
df
```

```python
grp=df.groupby("players")["runs"].sum()
print(grp)
type(grp)
```

Aggregation

```python
sum=df.groupby("players").agg(
    Total_runs=("runs","sum"),
    Average_runs=("runs","mean"),
    Matches_played=("matches","sum"),
    Highest_score=("runs","max"),
    Lowest_score=("runs","min"),
    Totals_balls=("Balls","sum"),
    Fifties=("runs",lambda x:((x>=50)&(x<100)).sum())
).reset_index()
sum
```
